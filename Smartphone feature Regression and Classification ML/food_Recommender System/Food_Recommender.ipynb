## STEP 1: Importing Libraries and Dependencies

```python
import pandas as pd
import numpy as np
from surprise import Reader, Dataset, KNNBasic, SVD, KNNWithZScore
from surprise.model_selection import train_test_split
from surprise import accuracy
```

## step 2: Reading our Data

```python
food =pd.read_csv(r"C:\Users\chukw\Documents\RAIN\AIML 2ND SEMESTER (File responses)\AIML 211 Exam\food_dataset\food_dataset\Food_data.csv")
ratings = pd.read_csv(r"C:\Users\chukw\Documents\RAIN\AIML 2ND SEMESTER (File responses)\AIML 211 Exam\food_dataset\food_dataset\ratings.csv")
```

```python
food.head(10)
```

```python
ratings.head(10)
```

## Step 3: Preparing our Data for Suprise

```python
#Surprise needs a format of (userid,itemid,rating)
df = ratings[['User_ID', 'Food_ID', 'Rating']]
```

```python
df
```

#### Use reader assign the rating scale

```python
# GETTING THE UNIQUE RANGE
df["Rating"].unique()
```

##### since surprise is designed to work with sparse matrices, we can leave the NaN variables

```python
reader = Reader(rating_scale=(1, 10))
```

#### Load data into Surprise's Dataset format

```python
dataset = Dataset.load_from_df(df, reader)
```

## Step 4: Split Data into Train and Test Data

```python
trainset, testset = train_test_split(dataset, test_size=0.25, random_state=100)
```

```python
trainset
```

```python
testset
```

## Step 5: Train Model using KNNBasic(User-Based)

```python
# Build KNNBasic model with Pearson Correlation between users
model = KNNBasic(k=40,min_k=3,
                sim_options={"name":"pearson",
                            "user_based":True,
                            "min_support": 10})
```

```python
model.fit(trainset)
```

#### Checking the prediction on a random set

```python
model.predict(2,8)
```

```python
model.predict(2,8).est
```

## Step 6: Testing on the Entire Dataset

```python
predictions=model.test(testset)
predictions
```

## Step 7: Accuracy score with RMSE(Root-Mean-Square-Error)

```python
print('KNNBasic Model Performance:')
KNNrmse = accuracy.rmse(predictions)
```

# Step 8: Trying with different Models

## Trying with SVD

```python
svd_model = SVD(n_factors=50, n_epochs=20, random_state=42)


svd_model.fit(trainset)
```

```python
svd_predictions = svd_model.test(testset)
```

```python
print('SVD Model Performance:')
svd_rmse = accuracy.rmse(svd_predictions)
```

## Trying with KNNWithZScore

```python
model3= KNNWithZScore(k=40,min_k=3,
                sim_options={"name":"pearson",
                            "user_based":True,
                            "min_support": 10})
```

```python
model3.fit(trainset)
```

```python
predictionsKNNZ=model3.test(testset)
predictionsKNNZ
```

### A lot of the values contain details={'was_impossible': True, 'reason': 'User and/or item is unknown.'} issue

```python
print('KNNWithZScore Model Performance:')
KNNZ = accuracy.rmse(predictionsKNNZ)
```
