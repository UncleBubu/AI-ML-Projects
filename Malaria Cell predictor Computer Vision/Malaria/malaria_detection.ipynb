```python
X = []
y= []

#parasitized = 1
#uninfected =0

for filename in os.listdir(uninfected)[:5000]:
    try:
        img = cv2.imread(f"{uninfected}/{filename}")
        img = cv2.resize(img,(60,60))
        img = img/255 # normalizes the pixel by dividing by 255

        X.append(img)
        y.append(0)
    except:
        pass
```

```python
for filename in os.listdir(parasatised)[:5000]:
    try:
        img = cv2.imread(f"{parasatised}/{filename}")
        img = cv2.resize(img,(60,60))
        img = img/255

        X.append(img)
        y.append(1)
    except:
        pass
    
```

```python
X_arr = np.array(X)
y_arr = np.array(y)
```

```python
X_arr.shape
```

```python
y_arr.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_arr, y_arr, test_size = 0.3, random_state= 9)
```

```python
m = mobilenet_v2.MobileNetV2(input_shape = (60,60,3),include_top = False)

m.summary()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
model = Sequential()
model.add(m)  # assuming 'm' is a Conv2D or similar layer
model.add(Flatten())
model.add(Dense(128, activation="relu"))
model.add(Dense(128, activation="relu"))
model.add(Dense(256, activation="relu"))
model.add(Dense(1, activation="sigmoid"))
```

```python
model.compile(loss = "binary_crossentropy", optimizer = "adam", metrics = ["accuracy"])
```

```python
model.fit(X_train,y_train,epochs =30,verbose = 1, validation_split = 0.25)
```

```python
model.save("malaria_model.h5")
```
