```python
from dotenv import load_dotenv
import os
from langchain_core.prompts import PromptTemplate
from langchain_groq import ChatGroq
```

```python
load_dotenv()
```

```python
template = """ 
You are a biology teacher for secondary school students. You are great at explaining concepts in such a way that students understand easily. Always keep your explanations simple and short, dont forget to always ask the student if they understand. please explain photosynthesis.
"""
```

```python

prompt = PromptTemplate.from_template(template)
```

```python
llm = ChatGroq(model = "meta-llama/llama-4-scout-17b-16e-instruct",
temperature = 0.3,
api_key = os.environ.get("GROQ_API_KEY"))
```

```python
chain = prompt | llm
```

```python
prompt
```

```python
response = chain.invoke(input = {})
```

```python
response
```

```python
print(response.content)
```

## DYNAMIC PROMPTING

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
template = """ 
You are a biology teacher for secondary school students. You are great at explaining concepts in such a way that students understand easily. Always keep your explanations simple and short, dont forget to always ask the student if they understand. please explain this topic: {topic}.
"""
```

```python
prompt = PromptTemplate.from_template(template)
```

```python
chain = prompt | llm | StrOutputParser()
```

```python
response = chain.invoke(input = {"topic":"Evolution"})
```

```python
print(response)
```
